```python
class Animal:
    def __init__(self, name, age):
        self.name = name
        self.age = age
    def talk(self):
        return f'{self.name} talks'
    def __str__(self):
        return f'{self.name} and {self.age}'   
```

```python
a = Animal("tom", 12)
print(a)
```
