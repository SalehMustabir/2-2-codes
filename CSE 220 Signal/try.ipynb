```python
import matplotlib.pyplot as plt
```

```python
# First graph
xs = np.array([1,2,3,7,1])
ys = np.array([12,32,12,32,12])

plt.subplot(2,1,1)
plt.plot(xs, ys, marker='o', linestyle='--', color='red')
plt.title("First Plot")
plt.xlabel("x s")
plt.ylabel("y s")

# Second graph
xs = np.array([1,2,3,7,1])
ys = np.array([1,2,12,32,12])

plt.subplot(2,1,2)
plt.plot(xs, ys, marker='o', linestyle='--', color='blue')
plt.title("Second Plot")

plt.xlabel("x s")
plt.ylabel("y s")

plt.tight_layout()
plt.show()
```

```python
sizes = np.array([n*10 for n in range(1,6)])
print(sizes)
plt.scatter(xs, ys, s = sizes)
for x, y in zip(xs, ys):
    plt.annotate(f'({x}, {y})', (x, y),
                 textcoords="offset points",
                 xytext=(5,5),
                 fontsize=9)


plt.show()
```

```python
plt.bar(xs,ys)
plt.show
```

```python
plt.barh(xs,ys)
plt.show
```

```python
labels = [1,2,3,4,5]
plt.subplot(2,1,1)
plt.pie(xs, labels= labels)
plt.subplot(2,1,2)
plt.pie(ys)
plt.show()
```
